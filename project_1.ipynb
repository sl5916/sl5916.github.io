

# Project 1: Exploring the Distribution of Diamond Prices


## Introduction

When buying a diamond, does the average price really tell us how much a typical diamond costs?

In this project, I will explore the distribution of diamond prices using a dataset containing more than 50,000 diamonds. I am particularly interested in whether expensive diamonds pull the average price upward, making it less representative of what most diamonds cost.

I will calculate the mean, median, and mode using pandas and then repeat the calculations using only the Python standard library. Finally, I will create a simple visualization to better understand the distribution of diamond prices.

## Dataset

I am using the Diamonds dataset, which contains information on 53,940 diamonds, including their prices, carat weights, cuts, colors, and clarity grades.

Dataset source: https://github.com/mwaskom/seaborn-data/blob/master/diamonds.csv

For this project, I will focus on the `price` column, which records diamond prices in US dollars.



```python

import pandas as pd

df = pd.read_csv("diamonds.csv")

df.head()

```

## Part 1: Calculating Statistics with Pandas

To understand the distribution of diamond prices, I will calculate three measures of central tendency: mean, median, and mode.

The mean shows the average price of all diamonds, the median represents the middle price, and the mode identifies the most frequently occurring price.

Comparing these measures can help me understand whether the price distribution is symmetric or skewed.

```python
mean_price = df["price"].mean()
median_price = df["price"].median()
mode_price = df["price"].mode()

print("Mean:", mean_price)
print("Median:", median_price)
print("Mode:", mode_price)
```

### Interpretation

The mean diamond price is approximately USD 3,932.80, while the median is only USD 2,401. The most frequently occurring price is USD 605.

The mean is substantially higher than the median, suggesting that the distribution of diamond prices is right-skewed. Some expensive diamonds pull the average upward.

This means that the median may be a better measure of the price of a typical diamond, since it is less affected by extremely high prices.



## Part 2: Calculating Statistics Using the Python Standard Library

In this section, I will calculate the mean, median, and mode again without using pandas or built-in statistical functions.

I will read the original CSV file separately and use basic Python operations to perform the calculations. This will allow me to compare the results with those obtained using pandas.


```python
prices = []

with open("diamonds.csv", "r") as file:
    lines = file.readlines()

for line in lines[1:]:
    values = line.split(",")
    prices.append(int(values[6]))

print(len(prices))
```

```python
mean = sum(prices) / len(prices)

print("Mean:", mean)
```

```python
sorted_prices = sorted(prices)

n = len(sorted_prices)

if n % 2 == 0:
    median = (sorted_prices[n // 2 - 1] + sorted_prices[n // 2]) / 2
else:
    median = sorted_prices[n // 2]

print("Median:", median)
```

```python
price_counts = {}

for price in prices:
    if price in price_counts:
        price_counts[price] += 1
    else:
        price_counts[price] = 1

max_count = max(price_counts.values())

mode = []

for price in price_counts:
    if price_counts[price] == max_count:
        mode.append(price)

print("Mode:", mode)
```

### Comparing the Two Methods

Both methods produced the same results. The mean diamond price is approximately USD 3,932.80, the median is USD 2,401, and the mode is USD 605.

Pandas makes these calculations easier by providing built-in statistical functions. Using basic Python operations requires more steps, especially for calculating the mode, where I used a dictionary to count how often each price appeared.

## Part 3: Data Visualization

To better understand the distribution of diamond prices, I will create a text-based bar chart using basic Python operations.

I will divide the diamonds into different price ranges and count how many diamonds fall into each range. This will help me see whether most diamonds are relatively inexpensive or expensive.


```python
price_ranges = {
    "Under $2,000": 0,
    "$2,000-$3,999": 0,
    "$4,000-$5,999": 0,
    "$6,000-$9,999": 0,
    "$10,000 and above": 0
}

for price in prices:
    if price < 2000:
        price_ranges["Under $2,000"] += 1
    elif price < 4000:
        price_ranges["$2,000-$3,999"] += 1
    elif price < 6000:
        price_ranges["$4,000-$5,999"] += 1
    elif price < 10000:
        price_ranges["$6,000-$9,999"] += 1
    else:
        price_ranges["$10,000 and above"] += 1

print(price_ranges)
```

```python

print("Distribution of Diamond Prices")
print("Each * represents approximately 500 diamonds")
print()

for price_range in price_ranges:
    count = price_ranges[price_range]
    bar = "*" * (count // 500)
    print(price_range, ":", bar, count)
```


### Visualization Analysis

The text-based bar chart shows that diamond prices are not evenly distributed across the different price ranges.

A large number of diamonds are concentrated in the lower price ranges, while relatively fewer diamonds fall into the highest price category.

This pattern is consistent with the difference between the mean and median prices. Although the average diamond costs approximately USD 3,933 in this dataset, the median price is only USD 2,401.

Therefore, the median appears to be a more representative measure of a typical diamond's price than the mean.
